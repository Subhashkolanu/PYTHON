GUI using Tkinter

```python
import tkinter as s
root=s.Tk()
root.title("First GUI")
root.geometry("1920x1080")
root.mainloop()
```

```python
r=s.Tk()
r.title("Welcome screen")
label=s.Label(r,text="Say my name!",font=("Time new roman",16,"bold"),fg="white",bg="black")
label.pack(pady=20)
r.mainloop()
```

```python
#Changing background
from tkinter import *
r=Tk()
r.title("Background")
r.geometry("500x500")
r.configure(bg="BLUE")
mainloop()
```

```python
from tkinter import *
r=Tk()
r.title("Label")
r.geometry("500x500")
d=Label(r,"Simply men")
d.place(x=20,y=20)
```
